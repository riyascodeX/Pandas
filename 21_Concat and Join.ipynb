```python
import pandas as pd
```

# 1.CONCAT 

```python
data1=pd.DataFrame([[1,11],[2,12]],columns=['A','B'])
data2=pd.DataFrame([[1,11],[2,12]],columns=['A','B'])
```

```python
# SAME COLUMN-->MERGE
# NO SAME COLUMN-->NULL  
pd.concat([data1,data2])
```

# 2.JOIN

```python
data3=pd.DataFrame([[1,11],[2,12]],columns=['A','B'])

data4=pd.DataFrame([[1,11],[2,12]],columns=['D','B'])
```

```python
# SAME COLUMN --> ADD SUFFIX LEFT OR RIGHT TABLE
# NO SAME COLUMN --> MERGE
data3.join(data4,lsuffix='_left')
```

```python
# SAME COLUMN --> ADD SUFFIX LEFT OR RIGHT TABLE
# NO SAME COLUMN --> MERGE
data3.join(data4,rsuffix='_left')
```
