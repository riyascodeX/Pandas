# INTERMITENT ROW AND COLUMN DELETE METHOD

```python
# ONE ROW DELETE METHOD
a.drop(0,axis=0)
```

```python
# INTERMITENT ROW DELETE METHOD
a.drop([0,1,2,3],axis=0)
```

```python
# ONE COLUMN DELETE METHOD
a.drop('Car',axis=1)
```

```python
# INTERMITENT COLUMN DELETE METHOD
a.drop(columns=['Car','MPG'],axis=0)
```
