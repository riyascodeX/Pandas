# FUNCTIONS

```python
import pandas as pd
```

```python
a=pd.read_csv('data/cars.csv')
a
```

# 1.CUSTOM FUNCTION WITH SERIES ARGUMENTS

```python
# SERIES ARGUMENTS
def convert_one(word:str): # STR--> ONE COLUMN
    return word.lower().replace(" ","");
```

```python
# ARGUMENTS ALREADY (STRING DATA TYPES) BECUASE ONE COLUMN SO DIRECTLY APPLY
a['Car'].apply(convert_one) 
```

# 2.CUSTOM FUNCTION WITH DATA FRAMES SERIES ARGUMENTS

```python
# DATA FRAME SERIES ARGUMENTS
def convert_double(COLUMN:pd.Series): # PD.SERIES--> MORE COLUMN
    return COLUMN.str.lower().str.replace(" ",""); # EACH FUNCTION LIKE LOWER,UPPER USE **(.STR) BECAUSE DATAFRAME SERIES ARGUMENTS MORE COLUMN
```

```python
# SRORE COLUMNS IN VAIRABLE
dfq=a[['Car','Origin',]]
dfq
```

```python
# dfq --> TWO OR MORE COLUMN STORE ONE VARIABLE THEN USE THIS
dfq.apply(convert_double)
```

# 3.ASSIGN FUNCTION

```python
# FUNCTION
def columns_mdify_insert(col):
    return col*2
```

```python
# SYNTAX --> a.assign(new column | exixting column = function_name(column_name))
a.assign(Cylinders=columns_mdify_insert(a['Cylinders']))
```

# 5.GROUP BY FUNCTION

```python
# METHOD 1
a['Weight'].groupby(a['Origin']).max()
```

```python
# METHOD 2
a.groupby('Origin')['Weight'].max()
```

# 6.TRANSFORM FUNCTION

```python
# YOU CAN STORE AGGREGATE RESULT IN NEW COLUMN USE **TRANSFORM 
a['transform']=a.groupby('Origin')['Weight'].transform('max')
a 
```

# 7.MAP FUNCTION

```python
# COLUMN UNIQE VALUES ONLY DISPLAY
a['Origin'].unique()
```

```python
# EXISTING COLUMN NEW VALUE SET THEN NEW COLUMN ADD
a['map_set']=a['Origin'].map({
    'US':1,
    'Europe':2,
    'Japan':3
})
a
```

# 8.Lambda function

```python
# LAMBDA FUNCTION-->(lambda argument:expression)
a['Cylinders'].apply(lambda x:x-5)
```

# 9.PIPE FUNCTION

```python
# FUNCTION 1
def pipe1(col1):
    return col1*2.1755

# FUNCTION 2 
def pipe2(col2):
    return round(col2,2)
```

```python
# METHOD 1
a['a1']=a['Weight'].apply(pipe1)
a
```

```python
# METHOD 1
a['a1']=a['a1'].apply(pipe2)
a
```

```python
# METHOD 2 -->FUNCTION 1,2 THATS PIPE FUNCTION BENIFITS
a['pipe']=a['Weight'].pipe(pipe1).pipe(pipe2)
a
```
