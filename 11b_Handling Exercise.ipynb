# prefix and suffix

```python
# ADD PREFIX IN COLUMN
a.add_prefix('PREFIX_',axis=1)
```

```python
# ADD PREFIX IN ROW
a.add_prefix('PREFIX_',axis=0)
```

```python
# ADD SUFFIX IN ROW 
a.add_suffix('_SUFFIX',axis=0)
```

```python
# ADD SUFFIX IN AOLUMN
a.add_suffix('_SUFFIX',axis=1)
```

# min and max

```python
# SPECIFIC COLUMN MINIMUM VALUES
a['MPG'].idxmin()
```

```python
# SPECIFIC COLUMN MAXIMUM VALUES
a['MPG'].idxmax()
```

```python
# SPECIFIC COLUMN MINIMUM **NUMPY VALUES
a['MPG'].argmin()
```

```python
# SPECIFIC COLUMN MAXIMUM **NUMPY VALUES
a['MPG'].argmax()
```
