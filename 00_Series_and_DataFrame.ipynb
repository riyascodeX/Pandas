# 1.LIST DATAFRAME

```python
apple=['apple',150]
banana=['banana','70']
fruits=[apple,banana]
df=pd.DataFrame(fruits)
df.columns=('name','price')
df
```

# 2.TUPLE DATAFRAME

```python
apple=('apple',150)
banana=('banana','70')
fruits=(apple,banana)
df=pd.DataFrame(fruits)
df.columns=('name','price')
df
```

# 3.DICTIONARY DATAFRAME

```python
apple={'name':'apple','price':'150'}
banana={'name':'banana','price':'70'}
df=pd.DataFrame(apple,banana)
df
```

# 3.1 DICTIONARIES OF LIST DATAFRAME

```python
#  SYNTAX:
#   {
#       "Column1": [value1, value2],
#       "Column2": [value1, value2]
#   }

data1={
    'NAME':['riyas','sameer'],
    'AGE':[20,33]
}
pd.DataFrame(data1)
```

# 3.2 LIST OF DICTIONARIES DATAFRAME

```python
#  SYNTAX:
#   [
#       {"Column1": value, "Column2": value},
#       {"Column1": value, "Column2": value}
#   ]



data2=[
    {'NAME':'riyas','AGE':20},
    {'NAME':'sameer','AGE':22}
]
pd.DataFrame(data2)
```

# 3.3 LIST OF LISTS DATAFRAME

```python
data3=[['riyas',20],['sameer',22]]
pd.DataFrame(data3,columns=['Name','AGE'])
```

# 3.4 NUMPY ARRAY DATAFRAME

```python
# NUMPY PACKAGE INSTALL
data4=np.array([['riyas',20],['sameer',22]])
pd.DataFrame(data4,columns=['Name','AGE'])
```

# 3.5 DICTIONARY OF SERIES DATAFRAME

```python
# EACH COLUMN IS A SERIES
n=pd.Series(['riyas','sameer']) # SERIES 1
a=pd.Series([20,22]) # SERIES 2
pd.DataFrame(
    {
        'NAME':n,
        'AGE':a
    }
)
```
