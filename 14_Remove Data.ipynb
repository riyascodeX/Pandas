```python
import pandas as pd
```

```python
df=pd.read_csv('Dataset/cars.csv')
df
```

# 1. DELETE ROWS INDEX METHOD

```python
# ONE ROW DELETE METHOD (AXIS=0)
df.drop(0,axis=0)
```

# 1.1 DELETE ROWS INTERMITTENT METHOD

```python
# INTERMITENT ROW DELETE METHOD
df.drop([0,1,2,3,4],axis=0)
```

# 2. DELETE COLUMN 

```python
# ONE COLUMN DELETE METHOD (AXIS=1),COLUMN=['CAR']
#df.drop(column=['Car'],axis=1)
df.drop(['Car'],axis=1)
```

# 2.1 DELETE COLUMN INTERMITTENT METHOD

```python
# INTERMITENT COLUMN DELETE METHOD
df.drop(['MPG','Model','Origin'],axis=1)
```

 # 3. POP DELETE METHOD

```python
# (ctrl + X) ---> cut and stored one variable
cut_pop= df.pop('MPG')
```

# 3.1 CHECK POP

```python
cut_pop
```

# 4. DELETE ROW TRUNCATE METHOD

```python
# TRUNCATE RANGE WISE SELECT DELETE METHOD
# (ctrl + X) ---> cut and stored one variable
# RULE 1 : INDEX NO SUFFLE
cut_trunc_row= df.truncate(0,10,axis=0)
```

# 4.1 CHECK TRUNCATE ROW

```python
cut_trunc_row
```

# 5. DELETE COLUMN TRUNCATE METHOD

```python
# 1. Must Sort the columns alphabetically
sort = df.sort_index(axis=1)
sort
```

```python
# 2. Then Run the truncate function
# (ctrl + X) ---> cut and stored one variable
cut_trunc_col= sort.truncate(before='Model',after='Origin',axis=1)
```

# 5.1 CHECK TRUNCATE COLUMN

```python
cut_trunc_col
```
