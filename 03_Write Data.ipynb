```python
import pandas as pd
```

# CREATE DATAFRMAE

```python
# Data
data1=['riyas',20,'kpm','A Grade']
data2=['sameer',21,'apm','C Grade']
data3=['sajid',22,'tmm','B Grade']
data4=['amir',20,'mpm',]
```

```python
#create table [pd.Dataframe]
df=pd.DataFrame([data1,data2,data3,data4],columns=['Name','Age','Place','Grade'])
df
```

# 1. DF TO CSV FILE

```python
# convert to csv file
#sep=';/./,'
df.to_csv('students',index=False,sep='\t')
    
```

# 2. DF TO EXCEL FILE

```python
#convert to excel file 
df.to_excel('students.xlsx',index=False,na_rep='ilai',sheet_name='avlotha')
```

# 3. DF TO JSON FILE

```python
#convert to json file
df.to_json('studets.json',orient='records')
```

# 4. DF TO PICKLE FILE

```python
#convert to pickle file
#[ compression ='gzip',compression='xz',protocol=1,2,3,4,5 ]
df.to_pickle('students.pk',compression='gzip',protocol=5)
```
