# FILTERING DATA LOC [ ]

```python
import pandas as pd
```

```python
a=pd.read_csv('data/cars.csv')
a
```

```python
# CONDITION 
a['MPG']>=30
```

```python
# APPLY LOC FOR FILTERED DATA
a.loc[a['MPG']>=30]
```

```python
a.loc[ a['Origin']=='US',['Car','MPG']]
```

# FILTER METHOD

```python
# FILTER METHOD USING ROW
a.filter(items=[1,2,3],axis=0)
```

```python
# FILTER METHOD USING COLUMN
a.filter(items=['Car','MPG'],axis=1)
```
